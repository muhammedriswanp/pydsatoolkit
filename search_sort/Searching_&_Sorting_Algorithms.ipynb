```python
def binary_search(numbers, target):
  left = 0
  right = len(numbers) - 1

  while left <= right:
    mid = (left + right) // 2
    if numbers[mid] == target:
      return mid
    elif numbers[mid] < target:
      left = mid + 1
    else:
      right = mid - 1
  return -1
```

```python
numbers = [10, 20, 30, 40, 50, 60, 70]
```

```python
print(binary_search(numbers, 60))
```

```python
def bubble_sort(numbers):
  n = len(numbers)

  for i in range(n):
    for j in range(n - i - 1):
      if numbers[j] > numbers[j+1]:
        numbers[j], numbers[j + 1] = numbers[j + 1], numbers[j]

  return numbers
```

```python
print(bubble_sort([5, 2, 4, 1, 3]))
```

Selection Sort — Python **code**

```python
def selection_sort(numbers):
  n = len(numbers)

  for i in range(n):
    min_index = i

    for j in range(i+1, n):
      if numbers[j] < numbers[min_index]:
        min_index = j

    numbers[i], numbers[min_index] = numbers[min_index], numbers[i]

  return numbers
```

```python
print(selection_sort([5, 2, 4, 1, 3]))
```

```python
def insertion_sort(numbers):
  for i in range(1, len(numbers)):
    key = numbers[i]

    j = i - 1

    while j >= 0 and key < numbers[j]:
      numbers[j + 1] = numbers[j]
      j -= 1

    numbers[j + 1] = key
  return numbers
```

```python
print(insertion_sort([5, 2, 4, 1, 3]))
```

```python

import random
import time

numbers = [random.randint(1, 10000) for _ in range(1000)]

def benchmark(sort_function, data):
    arr = data.copy()

    start = time.perf_counter()
    sort_function(arr)
    end = time.perf_counter()

    return end - start

algorithms = {
    "Bubble Sort": bubble_sort,
    "Selection Sort": selection_sort,
    "Insertion Sort": insertion_sort,
    "Python sort()": lambda arr: arr.sort()
}

for name, sort_function in algorithms.items():
    elapsed = benchmark(sort_function, numbers)
    print(f"{name}: {elapsed:.6f} seconds")
```


# Searching and Sorting Algorithms

## Time and Space Complexity

| Algorithm | Best Time | Average Time | Worst Time | Space |
|---|---|---|---|---|
| Linear Search | O(1) | O(n) | O(n) | O(1) |
| Binary Search | O(1) | O(log n) | O(log n) | O(1) |
| Bubble Sort* | O(n) | O(n²) | O(n²) | O(1) |
| Selection Sort | O(n²) | O(n²) | O(n²) | O(1) |
| Insertion Sort | O(n) | O(n²) | O(n²) | O(1) |
| Python sort() | O(n) | O(n log n) | O(n log n) | O(n) |

*Bubble Sort's best case is O(n) when using an early-stop
optimization that detects when no swaps occur.

## Key Notes

- Binary Search requires sorted input.
- Linear Search works on sorted or unsorted input.
- Bubble, Selection, and Insertion Sort use O(1) auxiliary space.
- Python's built-in sort() is generally preferred in real projects.
- Big-O describes how time or memory requirements grow as
  the input size increases.
